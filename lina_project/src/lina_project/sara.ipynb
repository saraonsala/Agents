```python
from embedchain import App
from dotenv import load_dotenv
import os

load_dotenv()
from embedchain import App
from sentence_transformers import SentenceTransformer

# Skapa en lokal embedding-modell
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# Skapa Embedchain-appen
app = App(embedder=embedding_model, db="chroma")

# Lägg till en lokal textfil som kunskapskälla
app.add("file.txt")  

# Ställ en fråga baserat på datan
response = app.query("Vad står det i filen?")
print(response)
```

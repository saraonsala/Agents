```python
from embedchain import App
import os
from dotenv import load_dotenv
load_dotenv()

OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')
#os.getenv("OPENAI_API_KEY")
# Skapa en Embedchain-app
app = App()

# Lägg till en datakälla (ex. en webbsida)
app.add("https://en.wikipedia.org/wiki/Artificial_intelligence")

# Ställ en fråga baserat på datan
response = app.query("What is artificial intelligence?")
print(response)
```
